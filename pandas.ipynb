```python
emp = pd.read_excel("./employee.xlsx")
```

```python
emp.dropna(inplace=True)
```

```python
emp.to_excel("./employee.xlsx") 
```

```python
print(emp[emp["Name"] == "Neha Brown"])
```

```python
emp["Salary"] = (
    emp["Salary"]
    .astype(str)     
    .str.replace(r"[$, ]", "", regex=True)
)
```

```python
emp["Salary"] = pd.to_numeric(emp["Salary"], errors="coerce")
```

```python
emp.groupby("Department")["Salary"].sum()
```

```python
salary = emp.loc[emp["Name"] == "Neha Brown" , "Salary"]
```

```python
emp["Salary"] = emp["Salary"].apply(lambda x : x + 100)
```

```python
plt.plot(emp["Salary"])
plt.title("Employee Salaries")
plt.xlabel("Employee Index")
plt.ylabel("Salary")
plt.show()
```

```python
plt.plot(emp["Salary"] , marker="o")
plt.title("Employee Salaries")
plt.xlabel("Employee Index")
plt.ylabel("Salary")
plt.show()
```

```python
plt.plot(emp["Salary"] , marker="o")
plt.title("Employee Salaries")
plt.xlabel("Employee Index")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

```python
dept_salary = emp.groupby("Department")["Salary"].sum()
plt.bar(dept_salary.index , dept_salary.values)
plt.title("Total Salary By Department")
plt.xlabel("Department")
plt.ylabel("Total Salary")
plt.xticks(rotation=45)
plt.show()
```

```python
#Pie Charts 

plt.pie(dept_salary.values , labels=dept_salary.index , autopct="%1.1f%%" )

plt.title("Salary Distribution By Depratment")
```
